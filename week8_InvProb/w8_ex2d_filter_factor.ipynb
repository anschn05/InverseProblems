```python
# W8 Exercise 2c)
import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(suppress=True)

M = 5
alpha = np.array([10**(-i) for i in range(M)])
print(alpha)

T = 0.01
j = np.arange(1, 65)
mu = np.exp(-j**2 * T)

omega = 1 / np.exp( -T )**2
k_array = [1,10,100,1000]

# TSVD
f_TSVD = lambda x,a: np.where(x**2 > a, 1, 0)

for i in range(M):
    plt.plot(mu,f_TSVD(mu,alpha[i]),label=f"alpha={alpha[i]}")
plt.legend()
plt.xlabel("mu")
plt.ylabel("q($\\mu$)")
plt.title("TSVD")
plt.show()
# TIKHONOV
f_tikhonov = lambda x,a: x**2/(x**2 + a)

for i in range(M):
    plt.plot(mu,f_tikhonov(mu,alpha[i]),label=f"alpha={alpha[i]}")
plt.legend()
plt.xlabel("mu")
plt.ylabel("q($\\mu$)")
plt.title("Tikhonov")
plt.show()

# LANDWEBER
f_landweber = lambda mu,k: 1 - (1 - omega * mu**2)**k

for k in k_array:
    plt.plot(mu,f_landweber(mu,k),label=f"k={k}")
plt.legend()
plt.xlabel("mu")
plt.ylabel("q($\\mu$)")
plt.title("Landweber")
plt.show()
```
