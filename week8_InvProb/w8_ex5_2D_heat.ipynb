```python
import numpy as np
from scipy.fft import dstn, idstn
from matplotlib import pyplot as plt

def heat_forward(x, T):
    n1,n2 = x.shape

    k = np.arange(1, n1 + 1)[:, None]
    l = np.arange(1, n2 + 1)[None, :]

    mu = np.exp(-T * (k**2 + l**2))

    c = dstn(x, type=2, norm="ortho")
    return idstn(mu * c, type=2, norm="ortho")



T = 10e-3
n_array = np.array([64,128,256,512])
n = 64

s_grid = [(j-1/2)*np.pi/n for j in range(n)]

s1, s2 = np.meshgrid(s_grid, s_grid, indexing="ij")
s_mesh = np.stack((s1, s2), axis=-1)

x = lambda s1,s2: s1*s2
x_mesh = x(s_mesh)

H = np.zeros((n,n))

y = heat_forward(x_mesh)

print(y)

plt.plot(y)
plt.show()
```
