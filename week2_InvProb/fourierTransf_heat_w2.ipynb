```python
import numpy as np
from math import pi, sqrt, sin, exp
from scipy.fft import dst, idst
import matplotlib.pyplot as plt

# def fourierTransformation(f):

N = 100
h = 1/N
T = 0.01      # verstrichene Zeit der Waermeleitung
s = np.array([pi*h*i for i in range(N)])        # [0, pi]
t = np.array([pi*h*i for i in range(N)])        # [0, pi]

# Anfangstemperatur an Stelle s
x = lambda s: s**2
x_s = x(s)      # x ausgewertet an stützstellen s
# sinusTransformation
c = dst(x_s, type=2)

# Temperatur an Stelle t nach verstrichener Zeit T
y_t = np.array([
    sqrt(2/pi) * sum( c[n] * exp(-T * n**2)  * sin(n*t[i]) for n in range(1,N)) 
    for i in range(N)])

# plt.plot(s,x_s,label="x(s)")
# plt.plot(t,y_t,label="y(t)")
# plt.legend()
# plt.show()

# MATRIX A berechnen
I = np.eye(N)
F = dst(I, type=2, axis=0)      #sinustransformation-matrix

B = np.zeros((N,N))
for i in range(N):
    for n in range(1,N):
        B[i,n] = sqrt(2/pi) * exp(-T*n**2) * sin(n*t[i]) 
A = B@F
y_A = A @ x_s

# EIGENVALUES of A
eigenvalues = np.linalg.svd(A,compute_uv=False)
print("eigenvalues:\n",np.round(eigenvalues,2))
```
