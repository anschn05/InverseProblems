```python
import numpy as np

# Interval: [0,1]

N = 1000 #stebsize
h = 1/N

s_array = [i*h + h/2 for i in range(N)]
t_array = [s_i + h/2 for s_i in s_array]
weights = 1/N
```

```python
f = lambda x: x*x
g = lambda x: x*x*x/3

g_si = [g(s_i) for s_i in s_array]      #ergebnis werte
f_ti = [f(t_i) for t_i in t_array]
A = np.tril(np.full((N,N),weights))

g_num = A@f_ti
# print(np.round(g_num,2))
# print(np.round(g_si,2))

#vergleichen die numerische Lösung mit der analytischen: (je feiner die Stützstellen, desto genauer)
print(g_num[-1])
print(g_si[-1])



```
