```python
"""noch nichts gemacht, nur code von woche 2 herkopiert"""

import numpy as np
from math import pi, sqrt
import matplotlib.pyplot as plt

def k(t,s):
    if s<t:
        return 1/sqrt(pi*(t-s))
    else:
        return 0

N = 1000 #stepsize
h = 1/N

s = np.array([i*h + h/2 for i in range(N)])
t = np.array([i*h for i in range(1,N+1)])

A = np.zeros((N,N))
for i in range(N):
    for j in range(N):
        A[i,j] = h * k(t[i], s[j])

# FUNCTION
f1 = lambda s: s**2
Kf1 = A @ f1(t)

# compare funciton 1 with the analytical solution
Kf1_analytisch = lambda t: 16/(15*sqrt(pi))*t**(5/2)

# PLOT
plt.plot(s,f1(s),label="f1(s)")
plt.plot(t,Kf1,label="Kf1(t)")
plt.plot(t,Kf1_analytisch(t),label="Kf1_analytisch(t)")
plt.legend()
plt.show()

# EIGENVALUES
eigenvalues = np.linalg.svd(A,compute_uv=False)
# print(eigenvalues)

plt.semilogy(eigenvalues)
plt.xlabel("index")
plt.ylabel("eigenvalues")
plt.show()
```
