```python
import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(suppress=True)

N = 50
h = 1/N
b = 3
x = np.array([b*h*i for i in range(N)])     #interval: (0,b)
M = 5
alpha = np.array([10**(-i) for i in range(M)])
print(alpha)

f_TSVD = lambda x,a: x**2/(x**2 + a)

for i in range(M):
    plt.plot(x,f_TSVD(x,alpha[i]),label=f"alpha={alpha[i]}")
plt.legend()
plt.show()

f_tikhonov = lambda x,a: np.where(x**2 > a, 1, 0)
for i in range(M):
    plt.plot(x,f_tikhonov(x,alpha[i]),label=f"alpha={alpha[i]}")
plt.legend()
plt.show()
```
