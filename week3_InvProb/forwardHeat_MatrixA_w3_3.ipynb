```python
import numpy as np
from numpy import sqrt, sin, pi, exp
import matplotlib.pyplot as plt

N = 256
h = np.pi / N
T = 0.05
s_grid = [(i+1/2)*h for i in range(N)]

x = lambda s: s**2
x_grid = [x(s) for s in s_grid]

S = np.zeros((N,N))
for k in range(N):
    for j in range(N):
        S[k,j] = sqrt(2/pi) * sin(j*s_grid[k])

#c = S.T * x_grid
D = np.diag([exp(-T*j**2) for j in range(N)])

A = S @ D @ S.T * (2/N)

SV_A = np.linalg.svd(A,compute_uv=False)

y_grid = A @ x_grid

mu = [exp(-T*j**2) for j in range(N)]    # Singular values of K

plt.semilogy(mu[0:50],label="mu")
plt.semilogy(SV_A[0:50],label="SVs of A")
plt.legend()
plt.show()
```
